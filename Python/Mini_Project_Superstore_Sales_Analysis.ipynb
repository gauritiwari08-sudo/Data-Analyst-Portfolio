# Apply and Map

Practice:

Create a new column such as:

Profit Status ("Profit" / "Loss")

Sales Category ("Low", "Medium", "High")

Use map() to convert Region names to short codes (for example, West → W, East → E).

```python
print("Profit Status (Profit / Loss)")
df["Profit_Status"]=df["Profit"].apply(lambda x:"Profit" if x>0 else "Loss")
#df["Profit_Status"]=df["Profit"].apply(lambda x:"Profit" if x>0 else "Loss")
df[["Profit","Profit_Status"]]
```

```python
print("Sales Category (Low, Medium, High)")
df["Sales_Category"]=df["Sales"].apply(lambda x:"Low" if x<1000 else ("Medium" if x<5000 else "High"))
df[["Sales","Sales_Category"]]
```

```python
print("Use map() to convert Region names to short codes (for example, West → W, East → E).")
df["Region_Code"]=df["Region"].map({"West":"W","East":"E","Central":"C","South":"S"})
df[["Region","Region_Code"]]
```
