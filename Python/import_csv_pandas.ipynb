```python
# Filtering : Keeping the rows that match a condition

tall_pokemon = df[df["Height"]>=2]
#print(tall_pokemon)

heavy_pokemon = df[df["Weight"]>=200]
#print(heavy_pokemon)

water_pokemon = df[(df["Type1"]=="Water") | (df["Type2"]=="Water")  ]
#print(water_pokemon)

fire_Flying = df[(df["Type1"]=="Fire") & (df["Type2"]=="Flying")  ]
print(fire_Flying)
```
