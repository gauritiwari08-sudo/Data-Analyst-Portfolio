```python
from matplotlib import colors
import matplotlib.pyplot as plt
import pandas as pd

x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y, color="Red",linestyle="--", marker="x")
plt.title("Sample Graph")
plt.xlabel("X-axis")
plt.ylabel("Y axis")
plt.title("Sample Graph")
plt.legend(["Sample Graph"])
figure_size=(10,5)
plt.savefig("sample_graph.png")
plt.grid(True)
plt.show()
```

```python

data = {
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'],
    'Electronics_Sales': [12000, 15000, 17000, 16000, 18000, 20000, 22000, 21000, 19000, 23000, 25000, 27000],
    'Clothing_Sales': [9000, 11000, 12000, 11000, 13000, 14000, 15000, 16000, 14000, 17000, 19000, 22000]
}

df=pd.DataFrame(data)
#print(df)
plt.plot(df['Month'],df['Clothing_Sales'],marker='o')
plt.plot(df['Month'],df['Electronics_Sales'],marker='v')

plt.title("Sales Graph")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)
#plt.legend(loc='upper left')
plt.show()
```

Bar Chart

```python
a = [1, 2, 3, 4, 5]
b = [15, 10, 8, 16, 21]
plt.bar(a,b,color="yellow")
plt.show()
```

```python
import numpy as np
a = [1, 2, 3, 4, 5]
b = [15, 10, 8, 16, 21]
colors = plt.cm.viridis(np.linspace(0, 1, len(b)))

plt.barh(a,b,color=colors)
plt.show()
```

Scatter plot

```python
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]
plt.figure(figsize=(12, 4))

plt.scatter(x, y, color='blue', marker='x')

plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Scatter Plot')

plt.show()
```

pie chart


```python

labels = ["A", "B", "C", "D"]
values = [30, 25, 20, 25]

explode=[0,0,0,0.1]
plt.pie(values, labels=labels,explode=explode)
plt.title("Simple Pie Chart")
plt.show()
```

```python
import matplotlib.pyplot as plt

data = [40, 30, 20, 10]
my_labels = ['Apples', 'Bananas', 'Cherries', 'Dates']

# Capture the patches (the color blocks) from the pie chart
patches, texts = plt.pie(data)

# Move legend outside: bbox_to_anchor coordinates are (x, y)
# loc="center left" aligns the vertical center of the legend box to your y coordinate
plt.legend(patches, my_labels, title="Fruits", loc="center left", bbox_to_anchor=(1, 0.5))

# Optional: Adjust layout so the legend isn't cropped out
plt.tight_layout()
plt.show()
```

```python
import numpy as np
figure,axes=plt.subplots(2,2)
x=np.array([1,2,3,4,5])
axes[0,0].plot(x,x**2)
axes[0,0].set_title("x**2")

axes[0,1].plot(x,x**3,color="red",marker="o")
axes[0,1].set_title("x**3")

axes[1,0].bar(x,x**4,color="yellow")
axes[1,0].set_title("x**4")

axes[1,1].plot(x,x**5,color="orange")
axes[1,1].set_title("x**5",y=0.05,ha="center")
plt.tight_layout()
plt.show()
```
