#Pandas Library

```python
import pandas as pd
#Series
data = [100,102,104]
#creating series with default series no
series =pd.Series(data)
print(series)
#change index to e101
series = pd.Series(data,index=['e101','e102','e103'])
print(series)
#accesing elecmetnt by its location
print(series.loc['e101']) #location by label
print(series.iloc[1]) #integer location
#change with loc
series.loc['e101']=105
print(series)
#filtering
print(series[series<103])

```

Dataframe same as excel tablw

```python
data={"Name":["Nabhya","Naksh","Nivesh"],
      "Age":["23","34","45"]
     }
df=pd.DataFrame(data)
print(df)

df=pd.DataFrame(data,index=["e101","e102","e103"])
print(df)

#select a single row loc property
print(df.loc["e101"])

#select a single row by using interger location
print(df.iloc[1])

#add new coloumn
df["Job"]=["Collector","IAS","IPS"]
print(df)

#add new row
new_row=pd.DataFrame([{"Name":"Ruhee","Age":"36","Job":"Scientist"}],index=["e104"])
df=pd.concat([df,new_row])
print(df)
```
